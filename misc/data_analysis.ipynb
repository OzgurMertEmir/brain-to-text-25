```python
from utils.load_data import load_h5py_file
from omegaconf import OmegaConf
import pandas as pd

csv_path = '../data/t15_copyTaskData_description.csv'
data_path = '../data/hdf5_data_final'
data_dir = '../data/hdf5_data_final'
model_path = '../data/t15_pretrained_rnn_baseline'
eval_type = 'train' # 'test', 'val', 'train'

model_args = OmegaConf.load(os.path.join(model_path, 'checkpoint/args.yaml'))

b2txt_csv_df = pd.read_csv(csv_path)
```

```python
model_data = {
    'train': {},
    'test': {},
    'val': {}
}
total_trials = {
    'train': 0,
    'test': 0,
    'val': 0
}
for session in model_args['dataset']['sessions']:
    for eval_type in ['train', 'val', 'test']:
        files = [f for f in os.listdir(os.path.join(data_dir, session)) if f.endswith('.hdf5')]
        if f'data_{eval_type}.hdf5' in files:
            eval_file = os.path.join(data_dir, session, f'data_{eval_type}.hdf5')

            data = load_h5py_file(eval_file, b2txt_csv_df)
            model_data[eval_type][session] = data

            total_trials[eval_type] += len(model_data[eval_type][session]["neural_features"])
            print(f'Loaded {len(model_data[eval_type][session]["neural_features"])} {eval_type} trials for session {session}.')

for eval_type in ['train', 'val', 'test']:
    print(f'Total number of {eval_type} trials: {total_trials[eval_type]}')
```

- The session date, block number, and trial number
- 512 neural features (2 features [-4.5 RMS threshold crossings and spike band power] per electrode, 256 electrodes), binned at 20 ms resolution. The data were recorded from the speech motor cortex via four high-density microelectrode arrays (64 electrodes each). The 512 features are ordered as follows in all data files:
    - 0-64: ventral 6v threshold crossings
    - 65-128: area 4 threshold crossings
    - 129-192: 55b threshold crossings
    - 193-256: dorsal 6v threshold crossings
    - 257-320: ventral 6v spike band power
    - 321-384: area 4 spike band power
    - 385-448: 55b spike band power
    - 449-512: dorsal 6v spike band power
- The ground truth sentence label
- The ground truth phoneme sequence label

```python
train_sentences = []
test_senteneces = []

for session in model_data['train']:
    train_sentences.extend(model_data['train'][session]['sentence_label'])
for session in model_data['val']:
    train_sentences.extend(model_data['val'][session]['sentence_label'])

for session in model_data['test']:
    test_senteneces.extend(model_data['test'][session]['sentence_label'])
```
